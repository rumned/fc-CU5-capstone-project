### From initial EDA we have found out that the districts are not consistent across datasets. Crime datasets refer to the IPD (Ibu Pejabat Polis Daerah) districts, education datasets refer to PPD (Pejabat Pendidikan Daerah) districts, while the rest refers to state territory districts. There are also aggregate districts included in the datasets, which sums the total stats for every district in the state. Spelling errors, inconsistent naming are also present. Federal territory names are simplified, districts are respelled and combined according to their geographical districts, and in some districts, combined according to their PPD districts to ensure data availability across columns. For the crime dataset, Putrajaya is included as one of Kuala Lumpur's districts, so we have to elevate the crime statistic to Putrajaya state

```python
# 1. Define Federal Territory Renames
wp_renames = {
    "W.P. Kuala Lumpur": "Kuala Lumpur",
    "W.P. Labuan": "Labuan",
    "W.P. Putrajaya": "Putrajaya"
}

# 2. Define Strategy 2 Mapping Dictionaries (Updated for stripped W.P. names)
safe_renames = {
    "Kulaijaya": "Kulai", "Ledang": "Tangkak", "Bandar Bharu": "Bandar Baharu",
    "Pasir Putih": "Pasir Puteh", "Cameron Highland": "Cameron Highlands",
    "Kuala Lipis": "Lipis", "Larut & Matang": "Larut dan Matang",
    "Larut Dan Matang": "Larut dan Matang","Larut Matang & Selama" : "Larut/Matang/Selama",
    "Manjung (Dinding)": "Manjung", "Jpn Perlis": "Perlis",
    "S.P. Selatan": "Seberang Perai Selatan", "Sp Selatan": "Seberang Perai Selatan",
    "S.P.Tengah": "Seberang Perai Tengah", "Sp Tengah": "Seberang Perai Tengah",
    "S.P.Utara": "Seberang Perai Utara", "Sp Utara": "Seberang Perai Utara",
    "Kota Kinabatangan": "Kinabatangan", "Pensiangan": "Nabawan",
    "Kota Samarahan": "Samarahan", "Lubok antu": "Lubok Antu", "Meradong": "Maradong",
    "Hulu Langat": "Ulu Langat", "Hulu Selangor": "Ulu Selangor", "Sg. Buloh": "Sungai Buloh",
    "Hulu": "Hulu Terengganu"
}

sub_districts_to_parents = {
    "Johor Bahru Selatan": "Johor Bahru", "Johor Bahru Utara": "Johor Bahru",
    "Iskandar Puteri": "Johor Bahru", "Nusajaya": "Johor Bahru",
    "Seri Alam": "Johor Bahru", "Pasir Gudang": "Johor Bahru",
    "Nilai": "Seremban", "Ipoh": "Kinta", "Batu Gajah": "Kinta",
    "Kinta Selatan": "Kinta", "Kinta Utara": "Kinta", "Sungai Siput": "Kuala Kangsar",
    "Gerik": "Hulu Perak", "Pengkalan Hulu": "Hulu Perak", "Tapah": "Batang Padang",
    "Tanjong Malim": "Muallim", "Taiping": "Larut dan Matang",
    "Arau": "Perlis", "Kangar": "Perlis", "Padang Besar": "Perlis",
    "Ampang Jaya": "Ulu Langat", "Kajang": "Ulu Langat", "Klang Selatan": "Klang",
    "Klang Utara": "Klang", "Petaling Jaya": "Petaling", "Subang Jaya": "Petaling",
    "Shah Alam": "Petaling", "Sungai Buloh": "Petaling", "Petaling Perdana": "Petaling",
    "Petaling Utama": "Petaling", "Brickfields": "Kuala Lumpur",
    "Cheras": "Kuala Lumpur", "Dang Wangi": "Kuala Lumpur",
    "Sentul": "Kuala Lumpur", "Wangsa Maju": "Kuala Lumpur"
}

fine_to_combined = {
    "Kuala Muda": "Kuala Muda/Yan", "Yan": "Kuala Muda/Yan",
    "Kulim": "Kulim/Bandar Baharu", "Bandar Baharu": "Kulim/Bandar Baharu",
    "Jempol": "Jempol/Jelebu", "Jelebu": "Jempol/Jelebu",
    "Larut dan Matang": "Larut/Matang/Selama", "Selama": "Larut/Matang/Selama",
    "Telupid": "Telupid/Tongod", "Tongod": "Telupid/Tongod",
    "Tatau": "Tatau/Sebauh", "Sebauh": "Tatau/Sebauh",
    "Matu": "Matu/Daro", "Daro": "Matu/Daro"," Matu Daro": "Matu/Daro"
}

# Combine maps
master_map = {**safe_renames, **sub_districts_to_parents, **fine_to_combined, **wp_renames}

print(master_map)
```

```python
# 3. Apply mappings and aggregate overlapping rows
for name, df in dfs.items():
    # Strip whitespace to prevent missed matches
    df['state'] = df['state'].str.strip()
    df['district'] = df['district'].str.strip()
    
    # Elevate Putrajaya and Labuan out of any nested state (fixes the crime dataset)
    df.loc[df['district'].isin(['W.P. Putrajaya', 'Putrajaya']), 'state'] = 'Putrajaya'
    df.loc[df['district'].isin(['W.P. Labuan', 'Labuan']), 'state'] = 'Labuan'
    
    # Strip 'W.P. ' from all State and District labels
    df['state'] = df['state'].replace(wp_renames)
    
    # Standardize sub-districts, fine units, and spelling
    df['district'] = df['district'].replace(master_map)
    
    # Isolate non-numeric columns to keep them distinct in the groupby
    group_cols = [col for col in ['state', 'district', 'date', 'category', 'type', 'stage', 'sex', 'age', 'ethnicity'] if col in df.columns]
    
    # Group by the new standardized labels to collapse sub-units mathematically
    dfs[name] = df.groupby(group_cols, as_index=False).sum(numeric_only=True)
```
